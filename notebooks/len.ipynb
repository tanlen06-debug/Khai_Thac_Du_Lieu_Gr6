## 0.3. Giới thiệu bộ dữ liệu Dry Bean

Dry Bean Dataset là bộ dữ liệu về các đặc trưng hình thái của hạt đậu, được công bố trên UCI Machine Learning Repository.

Bộ dữ liệu gồm:

### Các đặc trưng đầu vào

Bộ dữ liệu chứa 16 thuộc tính hình thái được trích xuất từ hình ảnh các hạt đậu:

1. **Area:** Diện tích vùng hạt đậu.
2. **Perimeter:** Chu vi đường bao hạt.
3. **MajorAxisLength:** Chiều dài trục chính.
4. **MinorAxisLength:** Chiều dài trục phụ.
5. **AspectRation:** Tỷ lệ giữa các trục.
6. **Eccentricity:** Độ lệch tâm.
7. **ConvexArea:** Diện tích vùng bao lồi.
8. **EquivDiameter:** Đường kính tương đương.
9. **Extent:** Tỷ lệ diện tích hạt trong vùng bao.
10. **Solidity:** Độ đặc của vùng hình dạng.
11. **roundness:** Độ tròn.
12. **Compactness:** Độ gọn hình dạng.
13. **ShapeFactor1:** Đặc trưng hình dạng số 1.
14. **ShapeFactor2:** Đặc trưng hình dạng số 2.
15. **ShapeFactor3:** Đặc trưng hình dạng số 3.
16. **ShapeFactor4:** Đặc trưng hình dạng số 4.

Các thuộc tính trên được sử dụng làm biến đầu vào cho những thuật toán khai thác dữ liệu.

## 0.4. Phát biểu bài toán

Từ bộ dữ liệu Dry Bean, nhóm đặt ra hai bài toán khai thác dữ liệu chính.

### Bài toán 1: Phân lớp các giống đậu

Cho tập dữ liệu gồm 16 đặc trưng hình thái của mỗi hạt đậu, xây dựng mô hình dự đoán hạt đậu thuộc một trong bảy giống đã biết.

Đầu vào:

- Ma trận đặc trưng X gồm 16 thuộc tính hình thái.

Đầu ra:

- Nhãn dự đoán của hạt đậu thuộc một trong bảy lớp.

Nhóm sử dụng ba thuật toán K-Nearest Neighbors (KNN), Random Forest và Support Vector Machine (SVM) để thực nghiệm.

Các mô hình được tối ưu siêu tham số và đánh giá bằng phương pháp 10-fold Cross-Validation, sử dụng F1-macro làm thước đo so sánh chính.

### Bài toán 2: Gom cụm các hạt đậu

Cho các đặc trưng hình thái của hạt đậu nhưng không sử dụng thông tin nhãn `Class` trong quá trình gom cụm, áp dụng các thuật toán để phát hiện những nhóm dữ liệu có đặc điểm tương đồng.

Hai thuật toán được sử dụng:

- K-Means.
- DBSCAN.

Kết quả gom cụm được đánh giá bằng Adjusted Rand Index (ARI), sử dụng nhãn thật để đối chiếu sau khi mô hình đã tạo cụm.

Qua đó, nhóm phân tích mức độ tương đồng giữa cấu trúc cụm được phát hiện và các giống đậu thực tế.

```python
# DATA-05: Phát hiện ngoại lệ bằng IQR

Q1 = numeric_features.quantile(0.25)
Q3 = numeric_features.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outlier_mask = (
    numeric_features.lt(lower) |
    numeric_features.gt(upper)
)

outlier_summary = pd.DataFrame({
    "Số giá trị ngoài IQR": outlier_mask.sum(),
    "Tỷ lệ (%)": (
        outlier_mask.mean() * 100
    ).round(2)
})

outlier_summary = outlier_summary.sort_values(
    "Số giá trị ngoài IQR",
    ascending=False
)

display(outlier_summary)

print(
    "Số dòng có ít nhất một ngoại lệ:",
    int(outlier_mask.any(axis=1).sum())
)

outlier_summary.to_csv(
    TABLES_DIR / "outlier_summary.csv",
    encoding="utf-8-sig"
)
```

```python
# DATA-07: Tìm các cặp tương quan mạnh nhất

# Chỉ lấy phần tam giác trên, bỏ đường chéo chính
upper = corr.where(
    np.triu(np.ones(corr.shape, dtype=bool), k=1)
)

# Chuyển ma trận thành bảng các cặp đặc trưng
pairs = upper.stack().reset_index()
pairs.columns = [
    "Đặc trưng 1",
    "Đặc trưng 2",
    "Pearson r"
]

# Loại các dòng thiếu hệ số tương quan
pairs = pairs.dropna(subset=["Pearson r"]).copy()

# Tính độ lớn tuyệt đối của hệ số tương quan
pairs["|r|"] = pairs["Pearson r"].abs()

# Sắp xếp từ tương quan mạnh nhất đến yếu nhất
pairs = pairs.sort_values(
    "|r|",
    ascending=False
).reset_index(drop=True)

print("5 cặp đặc trưng tương quan mạnh nhất:")
display(pairs.head(5))

# Kiểm tra bảng kết quả
print("Tổng số cặp tương quan:", len(pairs))
print("Số giá trị thiếu:", pairs["Pearson r"].isna().sum())

# Lưu bảng CSV
pairs.to_csv(
    TABLES_DIR / "correlation_pairs.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Đã lưu correlation_pairs.csv thành công!")
```

```python
if RUN_FULL_TRAINING:
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.model_selection import GridSearchCV, StratifiedKFold
    from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix
    # ============================================================
    # 07. KNN - Pipeline + GridSearchCV + Stratified 10-Fold
    # ============================================================

    import pandas as pd
    import numpy as np

    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.model_selection import GridSearchCV, StratifiedKFold
    from sklearn.metrics import (
        accuracy_score,
        f1_score,
        classification_report,
        confusion_matrix
    )

    # 1. Khởi tạo Stratified 10-Fold
    cv_knn = StratifiedKFold(
        n_splits=10,
        shuffle=True,
        random_state=42
    )

    # 2. Xây dựng Pipeline
    knn_pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('knn', KNeighborsClassifier())
    ])

    # 3. Xây dựng lưới tham số
    param_grid_knn = {
        'knn__n_neighbors': [3, 5, 7, 9, 11, 13, 15],
        'knn__weights': ['uniform', 'distance'],
        'knn__metric': ['minkowski'],
        'knn__p': [1, 2]
    }

    # 4. GridSearchCV với tiêu chí F1-macro
    grid_knn = GridSearchCV(
        estimator=knn_pipeline,
        param_grid=param_grid_knn,
        scoring='f1_macro',
        cv=cv_knn,
        n_jobs=-1,
        verbose=1
    )

    # 5. Huấn luyện trên tập Train
    grid_knn.fit(X_train, y_train)

    # 6. Lấy mô hình tốt nhất
    best_knn = grid_knn.best_estimator_

    print("===== KẾT QUẢ TỐI ƯU KNN =====")
    print("Best parameters:", grid_knn.best_params_)
    print("Best CV F1-macro:", round(grid_knn.best_score_, 4))

    # 7. Đánh giá trên tập Test
    y_pred_knn = best_knn.predict(X_test)

    print("\n===== ĐÁNH GIÁ TRÊN TEST =====")
    print("Accuracy:", round(accuracy_score(y_test, y_pred_knn), 4))
    print("F1-macro:", round(f1_score(
        y_test, y_pred_knn, average='macro'
    ), 4))

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred_knn))

    # 8. Ma trận nhầm lẫn
    cm_knn = confusion_matrix(y_test, y_pred_knn)

    cm_knn_df = pd.DataFrame(
        cm_knn,
        index=best_knn.classes_,
        columns=best_knn.classes_
    )

    display(cm_knn_df)
    import os
    import pandas as pd

    # Tạo thư mục lưu kết quả
    TABLES_DIR.mkdir(parents=True, exist_ok=True)

    # Lưu Classification Report
    report_dict = classification_report(
        y_test,
        y_pred_knn,
        output_dict=True
    )

    pd.DataFrame(report_dict).transpose().to_csv(
        TABLES_DIR / "knn_classification_report.csv",
        encoding="utf-8-sig"
    )

    # Lưu kết quả tổng hợp
    knn_results = pd.DataFrame([{
        "Best_Params": str(grid_knn.best_params_),
        "Best_CV_F1_Macro": grid_knn.best_score_,
        "Test_Accuracy": accuracy_score(y_test, y_pred_knn),
        "Test_F1_Macro": f1_score(
            y_test, y_pred_knn, average="macro"
        )
    }])

    knn_results.to_csv(
        TABLES_DIR / "knn_results.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("Đã lưu kết quả KNN thành công!")
    import matplotlib.pyplot as plt
    import seaborn as sns

    FIGURES_DIR.mkdir(parents=True, exist_ok=True)

    plt.figure(figsize=(9, 7))
    sns.heatmap(
        cm_knn,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=best_knn.classes_,
        yticklabels=best_knn.classes_
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title("KNN Confusion Matrix")

    plt.savefig(
        FIGURES_DIR / "knn_confusion_matrix.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
else:
    display(pd.read_csv(TABLES_DIR/"knn_results.csv",encoding="utf-8-sig"))
    print("Bảng KNN lấy từ output TV4 đã lưu. Chưa fit lại mô hình.")
```

## 05. SUPPORT VECTOR MACHINE (SVM) — TV4 BẢO

**Lý thuyết:** SVM tìm ranh giới quyết định tối đa hóa *margin*; các mẫu gần ranh giới nhất là *support vectors*. Kernel RBF cho phép học biên phi tuyến:

\[
K(x_i,x_j)=\exp(-\gamma\lVert x_i-x_j\rVert^2)
\]

`C` điều chỉnh mức phạt phân loại sai; `gamma` điều chỉnh mức ảnh hưởng của một điểm lên ranh giới kernel RBF.

```python
if RUN_FULL_CLUSTERING:
    # CLUST-02 & CLUST-03: Thực nghiệm K-Means (Elbow method và huấn luyện k=7)

    # 1. Khảo sát phương pháp Elbow với k từ 2 đến 10
    k_values = list(range(2, 11))
    inertias = []

    print("Đang chạy khảo sát Elbow Method cho K-Means (k=2..10)...")
    for k in k_values:
        km_temp = KMeans(n_clusters=k, init='k-means++', n_init=10, max_iter=300, random_state=RANDOM_STATE)
        km_temp.fit(X_scaled)
        inertias.append(km_temp.inertia_)

    # Vẽ biểu đồ Elbow
    plt.figure(figsize=(9, 5))
    plt.plot(k_values, inertias, marker='o', color='#1f77b4', linewidth=2.2, markersize=7)
    plt.axvline(x=7, color='#d62728', linestyle='--', linewidth=1.5, label='k=7 (Số giống đậu đề tài)')
    plt.title('Phương pháp Elbow Method: Đánh giá Inertia qua các giá trị k', fontsize=12, fontweight='bold', pad=12)
    plt.xlabel('Số lượng cụm k', fontsize=11)
    plt.ylabel('Inertia (Within-Cluster Sum of Squares)', fontsize=11)
    plt.xticks(k_values)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(frameon=True, facecolor='white', loc='upper right')
    plt.tight_layout()
    elbow_fig_path = FIGURES_DIR / "kmeans_elbow_inertia.png"
    plt.savefig(elbow_fig_path, dpi=300)
    plt.show()
    print(f"Biểu đồ Elbow đã được lưu tại: {elbow_fig_path}")

    # 2. Huấn luyện mô hình K-Means với k=7 cố định
    kmeans_7 = KMeans(
        n_clusters=7,
        init='k-means++',
        n_init=10,
        max_iter=300,
        random_state=RANDOM_STATE
    )
    km_labels = kmeans_7.fit_predict(X_scaled)
    km_inertia = kmeans_7.inertia_

    print(f"\nKết quả huấn luyện K-Means (k=7):")
    print(f"- Giá trị Inertia (WCSS): {km_inertia:,.2f}")

    # 3. Thống kê số lượng mẫu và tỷ lệ từng cụm
    km_counts = pd.Series(km_labels).value_counts().sort_index()
    km_summary = []

    for c in range(7):
        c_mask = (km_labels == c)
        c_size = c_mask.sum()
        pct = (c_size / len(y_true)) * 100
        top_class = y_true[c_mask].value_counts().index[0]
        top_class_cnt = y_true[c_mask].value_counts().iloc[0]
        purity = (top_class_cnt / c_size) * 100
        km_summary.append({
            'Cụm': f'Cluster {c}',
            'Số lượng mẫu': c_size,
            'Tỷ lệ (%)': round(pct, 2),
            'Giống đậu chiếm đa số': top_class,
            'Số mẫu giống chính': top_class_cnt,
            'Độ thuần khiết (Purity %)': round(purity, 2)
        })

    km_summary_df = pd.DataFrame(km_summary)
    km_summary_path = TABLES_DIR / "kmeans_cluster_summary.csv"
    km_summary_df.to_csv(km_summary_path, index=False)
    print(f"\nBảng tóm tắt cụm K-Means đã lưu tại: {km_summary_path}")
    print(km_summary_df.to_string(index=False))

    # 4. Lập bảng đối chiếu (Crosstab) giữa Cụm K-Means và Giống đậu thật
    crosstab_km = pd.crosstab(y_true, km_labels)
    crosstab_km.columns = [f'Cluster {c}' for c in crosstab_km.columns]
    crosstab_path = TABLES_DIR / "kmeans_crosstab_class.csv"
    crosstab_km.to_csv(crosstab_path)

    plt.figure(figsize=(10, 6))
    sns.heatmap(crosstab_km, annot=True, fmt='d', cmap='Blues', cbar=True, linewidths=0.5)
    plt.title('Ma trận đối chiếu giữa Cụm K-Means (k=7) và Nhãn Giống đậu Thực tế', fontsize=12, fontweight='bold', pad=12)
    plt.xlabel('Cụm dự đoán K-Means', fontsize=11)
    plt.ylabel('Giống đậu thực tế (Class)', fontsize=11)
    plt.tight_layout()
    heatmap_path = FIGURES_DIR / "kmeans_crosstab_heatmap.png"
    plt.savefig(heatmap_path, dpi=300)
    plt.show()
    print(f"Ma trận đối chiếu đã lưu tại: {heatmap_path}")
else:
    display(pd.read_csv(TABLES_DIR/"kmeans_cluster_summary.csv",encoding="utf-8-sig"))
    print("Bảng nhóm K-Means đọc từ output TV6 đã lưu.")
```

## 09. DBSCAN — TV6 THÀNH TÀI

**Nguyên lý:** DBSCAN gom cụm theo mật độ, không ấn định số cụm ban đầu. `eps` (ε) là bán kính lân cận; `min_samples` xác định số điểm tối thiểu để hình thành điểm lõi (core). Điểm biên nằm trong lân cận lõi; **điểm nhiễu mang nhãn −1**.

**Thực nghiệm TV6:** Đồ thị khoảng cách láng giềng thứ 32 làm cơ sở khảo sát. TV6 thử **27 cặp tham số**, với `eps` gồm 0.5,0.6,0.7,0.8,1.0,1.2,1.4,1.6,2.0 và `min_samples` gồm 16,32,50.


### Mã nguồn minh chứng — TV6 CLUST-04: Đồ thị 32-NN và Grid Search 27 cấu hình DBSCAN

*Trích từ `tai.ipynb, code cell 11` trên `main`. Những thay đổi giới hạn ở đường dẫn, khai báo dùng chung và chế độ chạy; **giữ nguyên thuật toán, lưới tham số, cách đánh giá**.*


```python
if RUN_FULL_CLUSTERING:
    # CLUST-04: Khảo sát k-distance và Grid Search tham số DBSCAN

    # 1. Tính toán k-NN Distance Graph (với k = 32)
    k_val = 32
    nn = NearestNeighbors(n_neighbors=k_val)
    nn.fit(X_scaled)
    distances, _ = nn.kneighbors(X_scaled)

    # Sắp xếp khoảng cách đến láng giềng thứ k tăng dần
    sorted_k_distances = np.sort(distances[:, -1])

    plt.figure(figsize=(9, 5))
    plt.plot(sorted_k_distances, color='#2ca02c', linewidth=2, label=f'Đường cong {k_val}-NN Distance')
    plt.axhline(y=0.8, color='#d62728', linestyle='--', linewidth=1.5, label='eps = 0.8 (Ngưỡng tối ưu ARI)')
    plt.axhline(y=1.0, color='#ff7f0e', linestyle=':', linewidth=1.5, label='eps = 1.0 (Ngưỡng bắt đầu bão hòa cụm)')
    plt.title(f'Đồ thị k-Distance Graph (k={k_val} = 2*m) để định hướng chọn eps', fontsize=12, fontweight='bold', pad=12)
    plt.xlabel('Chỉ số điểm dữ liệu (đã sắp xếp tăng dần)', fontsize=11)
    plt.ylabel(f'Khoảng cách đến láng giềng thứ {k_val}', fontsize=11)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(frameon=True, facecolor='white', loc='upper left')
    plt.tight_layout()
    kdist_fig_path = FIGURES_DIR / "dbscan_k_distance.png"
    plt.savefig(kdist_fig_path, dpi=300)
    plt.show()
    print(f"Biểu đồ k-distance đã lưu tại: {kdist_fig_path}")

    # 2. Khảo sát lưới Grid Search cho DBSCAN
    eps_grid = [0.5, 0.6, 0.7, 0.8, 1.0, 1.2, 1.4, 1.6, 2.0]
    min_samples_grid = [16, 32, 50]

    dbscan_results = []
    print("\nĐang tiến hành Grid Search cho DBSCAN...")

    for ms in min_samples_grid:
        for eps in eps_grid:
            db = DBSCAN(eps=eps, min_samples=ms)
            labels = db.fit_predict(X_scaled)

            # Số cụm thực tế (loại trừ nhãn nhiễu -1)
            unique_labels = set(labels)
            n_clusters = len(unique_labels) - (1 if -1 in unique_labels else 0)
            n_noise = (labels == -1).sum()
            noise_pct = (n_noise / len(labels)) * 100
            ari_score = adjusted_rand_score(y_true, labels)

            dbscan_results.append({
                'min_samples': ms,
                'eps': eps,
                'Số cụm (k)': n_clusters,
                'Số điểm nhiễu': n_noise,
                'Tỷ lệ nhiễu (%)': round(noise_pct, 2),
                'ARI': round(ari_score, 4)
            })

    dbscan_df = pd.DataFrame(dbscan_results)
    dbscan_table_path = TABLES_DIR / "dbscan_grid_search.csv"
    dbscan_df.to_csv(dbscan_table_path, index=False)
    print(f"Bảng kết quả Grid Search DBSCAN đã lưu tại: {dbscan_table_path}")

    print("\nTop 5 cấu hình có ARI cao nhất của DBSCAN:")
    top_dbscan = dbscan_df.sort_values(by='ARI', ascending=False).head(5)
    print(top_dbscan.to_string(index=False))

    print("\nCấu hình tiêu biểu thể hiện hiện tượng 'Cụm khổng lồ' (eps >= 1.0):")
    single_cluster_dbscan = dbscan_df[(dbscan_df['min_samples'] == 16) & (dbscan_df['eps'].isin([1.0, 1.2, 1.6]))]
    print(single_cluster_dbscan.to_string(index=False))
else:
    display(pd.read_csv(TABLES_DIR/"dbscan_grid_search.csv",encoding="utf-8-sig").sort_values("ARI",ascending=False).head(5))
```

```python
if RUN_FULL_CLUSTERING:
    # CLUST-05: Trực quan hóa và so sánh phân cụm trên không gian PCA 2D

    # 1. Tính toán PCA 2D trên dữ liệu đã chuẩn hóa
    pca_2d = PCA(n_components=2, random_state=RANDOM_STATE)
    X_pca = pca_2d.fit_transform(X_scaled)
    exp_var = pca_2d.explained_variance_ratio_

    pca_plot_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
    pca_plot_df['Class'] = y_true.values
    pca_plot_df['KMeans_Cluster'] = [f'Cluster {c}' for c in km_labels]

    # Dự đoán DBSCAN tối ưu (eps=0.8, min_samples=32)
    db_best = DBSCAN(eps=0.8, min_samples=32)
    db_best_labels = db_best.fit_predict(X_scaled)
    pca_plot_df['DBSCAN_Best'] = [f'Cluster {l}' if l != -1 else 'Nhiễu (-1)' for l in db_best_labels]

    # Dự đoán DBSCAN cụm lớn (eps=1.0, min_samples=16)
    db_dense = DBSCAN(eps=1.0, min_samples=16)
    db_dense_labels = db_dense.fit_predict(X_scaled)
    pca_plot_df['DBSCAN_Dense'] = [f'Cluster {l}' if l != -1 else 'Nhiễu (-1)' for l in db_dense_labels]

    # Tọa độ tâm cụm K-Means chiếu lên không gian PCA 2D
    centroids_pca = pca_2d.transform(kmeans_7.cluster_centers_)

    print(f"PCA explained variance ratio: PC1 = {exp_var[0]*100:.2f}%, PC2 = {exp_var[1]*100:.2f}%")
    print(f"Tổng phương sai giải thích: {sum(exp_var)*100:.2f}%")

    # 2. Vẽ đồ thị chi tiết K-Means trên PCA 2D
    plt.figure(figsize=(10, 7))
    sns.scatterplot(
        data=pca_plot_df, x='PC1', y='PC2', hue='KMeans_Cluster',
        palette='tab10', alpha=0.55, s=25, edgecolor=None
    )
    plt.scatter(
        centroids_pca[:, 0], centroids_pca[:, 1],
        c='black', marker='X', s=180, linewidths=2, edgecolors='white',
        label='Tâm cụm (Centroids)'
    )
    for idx, (cx, cy) in enumerate(centroids_pca):
        plt.annotate(
            f'C{idx}', (cx, cy), xytext=(6, 6), textcoords='offset points',
            fontsize=10, fontweight='bold', color='black',
            bbox=dict(boxstyle='round,pad=0.2', facecolor='yellow', alpha=0.8)
        )

    plt.title(f'K-Means Gom cụm (k=7, ARI = {adjusted_rand_score(y_true, km_labels):.4f}) trên không gian PCA 2D', fontsize=12, fontweight='bold', pad=12)
    plt.xlabel(f'PC1 ({exp_var[0]*100:.1f}% phương sai)', fontsize=11)
    plt.ylabel(f'PC2 ({exp_var[1]*100:.1f}% phương sai)', fontsize=11)
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True)
    plt.tight_layout()
    km_fig_path = FIGURES_DIR / "kmeans_clusters_pca.png"
    plt.savefig(km_fig_path, dpi=300)
    plt.show()
    print(f"Biểu đồ K-Means PCA đã lưu tại: {km_fig_path}")

    # 3. Vẽ đồ thị chi tiết DBSCAN trên PCA 2D (làm nổi bật điểm nhiễu)
    plt.figure(figsize=(10, 7))
    # Vẽ điểm nhiễu màu xám nhạt với ký hiệu dấu 'x'
    noise_mask = (pca_plot_df['DBSCAN_Best'] == 'Nhiễu (-1)')
    sns.scatterplot(
        data=pca_plot_df[noise_mask], x='PC1', y='PC2',
        color='#b0b0b0', alpha=0.35, s=20, marker='x', label='Điểm nhiễu (-1)'
    )
    # Vẽ các cụm mật độ hợp lệ
    sns.scatterplot(
        data=pca_plot_df[~noise_mask], x='PC1', y='PC2',
        hue='DBSCAN_Best', palette='tab10', alpha=0.65, s=28, edgecolor=None
    )
    plt.title(f'DBSCAN Gom cụm (eps=0.8, min_samples=32, ARI = {adjusted_rand_score(y_true, db_best_labels):.4f}) trên PCA 2D', fontsize=12, fontweight='bold', pad=12)
    plt.xlabel(f'PC1 ({exp_var[0]*100:.1f}% phương sai)', fontsize=11)
    plt.ylabel(f'PC2 ({exp_var[1]*100:.1f}% phương sai)', fontsize=11)
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True)
    plt.tight_layout()
    db_fig_path = FIGURES_DIR / "dbscan_clusters_pca.png"
    plt.savefig(db_fig_path, dpi=300)
    plt.show()
    print(f"Biểu đồ DBSCAN PCA đã lưu tại: {db_fig_path}")

    # 4. Vẽ bảng 4 đồ thị đối chiếu tổng thể (Comparison Grid)
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))

    # 4.1 Ground Truth (Nhãn thật)
    sns.scatterplot(
        data=pca_plot_df, x='PC1', y='PC2', hue='Class',
        palette='tab10', alpha=0.5, s=20, edgecolor=None, ax=axes[0, 0]
    )
    axes[0, 0].set_title('1. Nhãn thực tế (Ground Truth - 7 giống đậu)', fontsize=11, fontweight='bold')
    axes[0, 0].set_xlabel(f'PC1 ({exp_var[0]*100:.1f}%)')
    axes[0, 0].set_ylabel(f'PC2 ({exp_var[1]*100:.1f}%)')
    axes[0, 0].legend(loc='lower left', fontsize=8, frameon=True)

    # 4.2 K-Means (k=7)
    sns.scatterplot(
        data=pca_plot_df, x='PC1', y='PC2', hue='KMeans_Cluster',
        palette='tab10', alpha=0.5, s=20, edgecolor=None, ax=axes[0, 1]
    )
    axes[0, 1].scatter(
        centroids_pca[:, 0], centroids_pca[:, 1],
        c='black', marker='X', s=120, edgecolors='white', label='Centroids'
    )
    axes[0, 1].set_title(f'2. K-Means (k=7, ARI = {adjusted_rand_score(y_true, km_labels):.4f})', fontsize=11, fontweight='bold')
    axes[0, 1].set_xlabel(f'PC1 ({exp_var[0]*100:.1f}%)')
    axes[0, 1].set_ylabel(f'PC2 ({exp_var[1]*100:.1f}%)')
    axes[0, 1].legend(loc='lower left', fontsize=8, frameon=True)

    # 4.3 DBSCAN (eps=0.8, min_samples=32)
    sns.scatterplot(
        data=pca_plot_df[noise_mask], x='PC1', y='PC2',
        color='#b0b0b0', alpha=0.3, s=15, marker='x', label='Nhiễu (-1)', ax=axes[1, 0]
    )
    sns.scatterplot(
        data=pca_plot_df[~noise_mask], x='PC1', y='PC2',
        hue='DBSCAN_Best', palette='tab10', alpha=0.6, s=20, edgecolor=None, ax=axes[1, 0]
    )
    axes[1, 0].set_title(f'3. DBSCAN Tối ưu (eps=0.8, ms=32, ARI = {adjusted_rand_score(y_true, db_best_labels):.4f})', fontsize=11, fontweight='bold')
    axes[1, 0].set_xlabel(f'PC1 ({exp_var[0]*100:.1f}%)')
    axes[1, 0].set_ylabel(f'PC2 ({exp_var[1]*100:.1f}%)')
    axes[1, 0].legend(loc='lower left', fontsize=8, frameon=True)

    # 4.4 DBSCAN Bão hòa (eps=1.0, min_samples=16)
    dense_noise_mask = (pca_plot_df['DBSCAN_Dense'] == 'Nhiễu (-1)')
    palette_dense = {'Cluster 0': '#1f77b4', 'Cluster 1': '#ff7f0e', 'Nhiễu (-1)': '#b0b0b0'}
    sns.scatterplot(
        data=pca_plot_df[dense_noise_mask], x='PC1', y='PC2',
        color='#b0b0b0', alpha=0.3, s=15, marker='x', label='Nhiễu (-1)', ax=axes[1, 1]
    )
    sns.scatterplot(
        data=pca_plot_df[~dense_noise_mask], x='PC1', y='PC2',
        hue='DBSCAN_Dense', palette=palette_dense, alpha=0.5, s=20, edgecolor=None, ax=axes[1, 1]
    )
    axes[1, 1].set_title(f'4. DBSCAN Cụm khổng lồ (eps=1.0, ms=16, ARI = {adjusted_rand_score(y_true, db_dense_labels):.4f})', fontsize=11, fontweight='bold')
    axes[1, 1].set_xlabel(f'PC1 ({exp_var[0]*100:.1f}%)')
    axes[1, 1].set_ylabel(f'PC2 ({exp_var[1]*100:.1f}%)')
    axes[1, 1].legend(loc='lower left', fontsize=8, frameon=True)

    plt.tight_layout()
    comp_fig_path = FIGURES_DIR / "clustering_comparison_pca.png"
    plt.savefig(comp_fig_path, dpi=300)
    plt.show()
    print(f"Biểu đồ so sánh 4 khung hình đã lưu tại: {comp_fig_path}")
else:
    print("Hình tổng hợp gốc: outputs/figures/clustering_comparison_pca.png")
```
