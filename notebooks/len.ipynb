## 0.3. Giới thiệu bộ dữ liệu Dry Bean

Dry Bean Dataset là bộ dữ liệu về các đặc trưng hình thái của hạt đậu, được công bố trên UCI Machine Learning Repository.

Bộ dữ liệu gồm:

| Thành phần | Thông tin |
|---|---|
| Số lượng mẫu | 13.611 |
| Số đặc trưng đầu vào | 16 |
| Số cột nhãn | 1 |
| Tổng số cột | 17 |
| Số lớp | 7 |
| Loại bài toán | Phân lớp đa lớp và gom cụm |
| Biến mục tiêu | Class |

### Các đặc trưng đầu vào

Bộ dữ liệu chứa 16 thuộc tính hình thái được trích xuất từ hình ảnh các hạt đậu:

1. **Area:** Diện tích vùng hạt đậu.
2. **Perimeter:** Chu vi đường bao hạt.
3. **MajorAxisLength:** Chiều dài trục chính.
4. **MinorAxisLength:** Chiều dài trục phụ.
5. **AspectRation:** Tỷ lệ giữa các trục.
6. **Eccentricity:** Độ lệch tâm.
7. **ConvexArea:** Diện tích vùng bao lồi.
8. **EquivDiameter:** Đường kính tương đương.
9. **Extent:** Tỷ lệ diện tích hạt trong vùng bao.
10. **Solidity:** Độ đặc của vùng hình dạng.
11. **roundness:** Độ tròn.
12. **Compactness:** Độ gọn hình dạng.
13. **ShapeFactor1:** Đặc trưng hình dạng số 1.
14. **ShapeFactor2:** Đặc trưng hình dạng số 2.
15. **ShapeFactor3:** Đặc trưng hình dạng số 3.
16. **ShapeFactor4:** Đặc trưng hình dạng số 4.

Các thuộc tính trên được sử dụng làm biến đầu vào cho những thuật toán khai thác dữ liệu.

### Các lớp phân loại

Trong bài toán phân lớp, `Class` là nhãn mục tiêu mà mô hình cần dự đoán.

Trong bài toán gom cụm, cột `Class` không được sử dụng làm đầu vào huấn luyện. Nhãn thật chỉ được sử dụng sau quá trình gom cụm để đánh giá kết quả.

## 0.4. Phát biểu bài toán

Từ bộ dữ liệu Dry Bean, nhóm đặt ra hai bài toán khai thác dữ liệu chính.

### Bài toán 1: Phân lớp các giống đậu

Cho tập dữ liệu gồm 16 đặc trưng hình thái của mỗi hạt đậu, xây dựng mô hình dự đoán hạt đậu thuộc một trong bảy giống đã biết.

Đầu vào:

- Ma trận đặc trưng X gồm 16 thuộc tính hình thái.

Đầu ra:

- Nhãn dự đoán của hạt đậu thuộc một trong bảy lớp.

Nhóm sử dụng ba thuật toán K-Nearest Neighbors (KNN), Random Forest và Support Vector Machine (SVM) để thực nghiệm.

Các mô hình được tối ưu siêu tham số và đánh giá bằng phương pháp 10-fold Cross-Validation, sử dụng F1-macro làm thước đo so sánh chính.

### Bài toán 2: Gom cụm các hạt đậu

Cho các đặc trưng hình thái của hạt đậu nhưng không sử dụng thông tin nhãn `Class` trong quá trình gom cụm, áp dụng các thuật toán để phát hiện những nhóm dữ liệu có đặc điểm tương đồng.

Hai thuật toán được sử dụng:

- K-Means.
- DBSCAN.

Kết quả gom cụm được đánh giá bằng Adjusted Rand Index (ARI), sử dụng nhãn thật để đối chiếu sau khi mô hình đã tạo cụm.

Qua đó, nhóm phân tích mức độ tương đồng giữa cấu trúc cụm được phát hiện và các giống đậu thực tế.

## 06. Thiết kế thực nghiệm phân lớp

**Nhiệm vụ:** Thống nhất X, y, stratified train/test, 10-fold StratifiedKFold, scoring='f1_macro', random_state; chống rò rỉ dữ liệu bằng Pipeline.

**Bài viết của thành viên:**

- Mục tiêu và kiến thức liên quan: _Bổ sung_.
- Phương pháp, tham số và lý do lựa chọn: _Bổ sung_.
- Kết quả chạy thật (bảng/biểu đồ): _Bổ sung_.
- Nhận xét và hạn chế: _Bổ sung_.


```python

# ==========================================================
# CLS-01: GIAO THUC DANH GIA MO HINH PHAN LOP
# Nhom truong: Tan Len
# ==========================================================

import pandas as pd

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold
)

# 1. Tham so chung cho ca nhom
RANDOM_STATE = 42
TEST_SIZE = 0.20
N_SPLITS = 10
SCORING = "f1_macro"

# 2. Doc du lieu da xuat tu EDA cua Lan
data_path = PROCESSED_DIR / "dry_bean.csv"

if not data_path.exists():
    raise FileNotFoundError(
        f"Khong tim thay du lieu: {data_path}"
    )

df = pd.read_csv(data_path, encoding="utf-8-sig")

# 3. Tach dac trung va nhan
X = df.drop(columns=["Class"])
y = df["Class"].astype(str)

assert X.shape[1] == 16, "Can dung 16 dac trung."
assert y.nunique() == 7, "Can dung 7 lop."

if X.isna().any().any():
    raise ValueError("Du lieu co gia tri thieu.")

# 4. Chia train/test co giu ty le lop
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

# 5. Tao bo chia 10-fold chung
cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

# 6. In thong tin kiem tra
print("=== GIAO THUC DANH GIA ===")
print("Random state:", RANDOM_STATE)
print("Test size:", TEST_SIZE)
print("So folds:", cv.n_splits)
print("Scoring:", SCORING)

print("\n=== DU LIEU ===")
print("X:", X.shape)
print("y:", y.shape)
print("Train:", X_train.shape)
print("Test:", X_test.shape)

print("\n=== PHAN BO NHAN (%) ===")
distribution = pd.DataFrame({
    "Toan bo": y.value_counts(normalize=True) * 100,
    "Train": y_train.value_counts(normalize=True) * 100,
    "Test": y_test.value_counts(normalize=True) * 100
})

display(distribution.round(2))

print("\nCLS-01: Da thiet lap protocol.")
```
